###1. Randomized Quick Sort Implementation in C++

This C++ code provides an implementation of Randomized Quick Sort. It also includes a mechanism to count the number of comparisons made during the sorting process.

```python
%%writefile randomized_quick_sort.cpp
#include <iostream>
#include <vector>
#include <cstdlib> // For srand and rand
#include <ctime>   // For time

// Global variable to count comparisons
long long comparisons = 0;

// Function to swap two elements
void swap(int& a, int& b) {
    int temp = a;
    a = b;
    b = temp;
}

// Function to choose a random pivot and place it at the end
int randomPartition(std::vector<int>& arr, int low, int high) {
    // Seed the random number generator only once
    // This is generally handled better by seeding in main or once per program execution
    // For this example, if called multiple times, it might reseed.
    // A more robust solution would pass a random engine or seed once in main.
    // srand(time(0)); // Commented out to avoid re-seeding multiple times if function is called often

    int pivotIndex = low + rand() % (high - low + 1);
    swap(arr[pivotIndex], arr[high]); // Move pivot to end
    return arr[high]; // Return pivot value
}

// Partition function with chosen pivot
int partition(std::vector<int>& arr, int low, int high) {
    int pivot = randomPartition(arr, low, high); // Pivot is now at arr[high]
    int i = (low - 1);

    for (int j = low; j <= high - 1; j++) {
        comparisons++; // Increment comparison count for each element compared with pivot
        if (arr[j] <= pivot) {
            i++;
            swap(arr[i], arr[j]);
        }
    }
    swap(arr[i + 1], arr[high]);
    return (i + 1);
}

// Randomized Quick Sort function
void randomizedQuickSort(std::vector<int>& arr, int low, int high) {
    if (low < high) {
        // pi is partitioning index, arr[p] is now at right place
        int pi = partition(arr, low, high);

        // Separately sort elements before partition and after partition
        randomizedQuickSort(arr, low, pi - 1);
        randomizedQuickSort(arr, pi + 1, high);
    }
}

// Function to print an array
void printArray(const std::vector<int>& arr) {
    for (int x : arr) {
        std::cout << x << " ";
    }
    std::cout << std::endl;
}

int main() {
    // Seed the random number generator once
    srand(time(0));

    std::vector<int> arr = {10, 7, 8, 9, 1, 5};
    int n = arr.size();

    std::cout << "Original array: ";
    printArray(arr);

    comparisons = 0; // Reset comparison counter
    randomizedQuickSort(arr, 0, n - 1);

    std::cout << "Sorted array: ";
    printArray(arr);

    std::cout << "Number of comparisons: " << comparisons << std::endl;

    std::cout << "\nTesting with another array:\n";
    std::vector<int> arr2 = {12, 11, 13, 5, 6, 7};
    n = arr2.size();

    std::cout << "Original array: ";
    printArray(arr2);

    comparisons = 0; // Reset comparison counter
    randomizedQuickSort(arr2, 0, n - 1);

    std::cout << "Sorted array: ";
    printArray(arr2);

    std::cout << "Number of comparisons: " << comparisons << std::endl;

    return 0;
}
```

### Randomized Quick Sort Implementation in Python

This Python code provides an implementation of Randomized Quick Sort. It also includes a mechanism to count the number of comparisons made during the sorting process.

```python
import random

# Global variable to count comparisons
comparisons = 0

def swap(arr, i, j):
    arr[i], arr[j] = arr[j], arr[i]

def random_partition(arr, low, high):
    global comparisons
    pivot_index = random.randint(low, high)
    pivot_value = arr[pivot_index]
    swap(arr, pivot_index, high) # Move pivot to end

    i = low - 1
    for j in range(low, high):
        comparisons += 1 # Increment comparison count
        if arr[j] <= pivot_value:
            i += 1
            swap(arr, i, j)
    swap(arr, i + 1, high)
    return i + 1

def randomized_quick_sort(arr, low, high):
    if low < high:
        pi = random_partition(arr, low, high)
        randomized_quick_sort(arr, low, pi - 1)
        randomized_quick_sort(arr, pi + 1, high)

# Test cases
arr1 = [10, 7, 8, 9, 1, 5]
print(f"Original array 1: {arr1}")
comparisons = 0 # Reset comparison counter
randomized_quick_sort(arr1, 0, len(arr1) - 1)
print(f"Sorted array 1: {arr1}")
print(f"Number of comparisons for array 1: {comparisons}")

print("\nTesting with another array:")
arr2 = [12, 11, 13, 5, 6, 7]
print(f"Original array 2: {arr2}")
comparisons = 0 # Reset comparison counter
randomized_quick_sort(arr2, 0, len(arr2) - 1)
print(f"Sorted array 2: {arr2}")
print(f"Number of comparisons for array 2: {comparisons}")
```

### 2.Randomized Select Algorithm

The Randomized Select algorithm is used to find the k-th smallest (or i-th smallest) element in an unsorted array in expected linear time. It's an adaptation of the QuickSort partitioning logic.

The core idea is to pick a random pivot, partition the array around it, and then recursively search in the appropriate subarray based on the pivot's position relative to `i`.

```python
%%writefile randomized_select.cpp
#include <iostream>
#include <vector>
#include <cstdlib> // For srand and rand
#include <ctime>   // For time
#include <algorithm> // For std::swap

using namespace std; // Added using namespace std

// Function to swap two elements
void swap(int& a, int& b) {
    int temp = a;
    a = b;
    b = temp;
}

// Function to choose a random pivot and partition the array
// Returns the final index of the pivot element
int randomized_partition(vector<int>& arr, int low, int high) {
    // Choose a random pivot index
    int pivot_index = low + rand() % (high - low + 1);
    // Move pivot to the end
    swap(arr[pivot_index], arr[high]);
    int pivot = arr[high];

    int i = low - 1; // Index of smaller element
    for (int j = low; j <= high - 1; j++) {
        if (arr[j] <= pivot) {
            i++;
            swap(arr[i], arr[j]);
        }
    }
    swap(arr[i + 1], arr[high]);
    return (i + 1); // Return the partitioning index
}

// Recursive function to find the i-th smallest element
// 'i' is 1-based index (e.g., i=1 for the smallest, i=arr.size() for the largest)
int randomized_select(vector<int>& arr, int low, int high, int i) {
    // If the array contains only one element, return that element
    if (low == high) {
        return arr[low];
    }

    // Partition the array and get the pivot's final position
    int pi = randomized_partition(arr, low, high);

    // Calculate the number of elements in the left subarray + pivot itself
    // 'k' is the 1-based rank of the pivot within the current subarray [low...high]
    int k = pi - low + 1;

    // If pivot is the i-th smallest element
    if (i == k) {
        return arr[pi];
    }
    // If i is smaller than k, search in the left subarray
    else if (i < k) {
        return randomized_select(arr, low, pi - 1, i);
    }
    // If i is greater than k, search in the right subarray
    // We need to adjust 'i' because we are looking for the (i-k)-th smallest
    // element in the right subarray [pi+1...high].
    else {
        return randomized_select(arr, pi + 1, high, i - k);
    }
}

// Function to print an array (for verification, not part of select logic)
void printArray(const vector<int>& arr) {
    for (int x : arr) {
        cout << x << " ";
    }
    cout << endl;
}

int main() {
    // Seed the random number generator once
    srand(time(0));

    vector<int> arr1 = {3, 2, 1, 5, 4};
    int i1 = 3; // Find the 3rd smallest element
    cout << "Original array 1: ";
    printArray(arr1);
    // Pass a copy if you want to preserve the original array, as select modifies it
    vector<int> arr1_copy = arr1;
    int result1 = randomized_select(arr1_copy, 0, arr1_copy.size() - 1, i1);
    cout << "The " << i1 << "rd smallest element is: " << result1 << endl;

    cout << endl;

    vector<int> arr2 = {10, 7, 8, 9, 1, 5};
    int i2 = 2; // Find the 2nd smallest element
    cout << "Original array 2: ";
    printArray(arr2);
    vector<int> arr2_copy = arr2;
    int result2 = randomized_select(arr2_copy, 0, arr2_copy.size() - 1, i2);
    cout << "The " << i2 << "nd smallest element is: " << result2 << endl;

    cout << endl;

    vector<int> arr3 = {50, 20, 10, 40, 30};
    int i3 = 1; // Find the 1st smallest element
    cout << "Original array 3: ";
    printArray(arr3);
    vector<int> arr3_copy = arr3;
    int result3 = randomized_select(arr3_copy, 0, arr3_copy.size() - 1, i3);
    cout << "The " << i3 << "st smallest element is: " << result3 << endl;

    cout << endl;

    vector<int> arr4 = {100, 50, 70, 90, 80};
    int i4 = 5; // Find the 5th smallest element (largest)
    cout << "Original array 4: ";
    printArray(arr4);
    vector<int> arr4_copy = arr4;
    int result4 = randomized_select(arr4_copy, 0, arr4_copy.size() - 1, i4);
    cout << "The " << i4 << "th smallest element is: " << result4 << endl;

    return 0;
}
```

```python
def randomized_select(arr, low, high, i):
    # If the array contains only one element, return that element
    if low == high:
        return arr[low]

    # Partition the array and get the pivot's final position
    pi = random_partition_select(arr, low, high)

    # Calculate the number of elements in the left subarray + pivot itself
    k = pi - low + 1 # Position of pivot (1-based relative to current subarray)

    # If pivot is the i-th smallest element
    if i == k:
        return arr[pi]
    # If i is smaller than k, search in the left subarray
    elif i < k:
        return randomized_select(arr, low, pi - 1, i)
    # If i is greater than k, search in the right subarray
    # We need to adjust 'i' because we are looking for the (i-k)-th smallest
    # element in the right subarray, relative to that subarray's start.
    else:
        return randomized_select(arr, pi + 1, high, i - k)
```

### 3. Kruskal's Algorithm Implementation in C++

This C++ code provides an implementation of Kruskal's algorithm to find the Minimum Spanning Tree (MST) of a graph. It uses a Disjoint Set Union (DSU) data structure to efficiently detect cycles.

```python
%%writefile kruskals_algorithm.cpp
#include <iostream>
#include <vector>
#include <algorithm> // For std::sort
#include <numeric>   // For std::iota

// Structure to represent an edge in the graph
struct Edge {
    int u, v, weight;

    // Comparator for sorting edges by weight
    bool operator<(const Edge& other) const {
        return weight < other.weight;
    }
};

// Disjoint Set Union (DSU) class with Path Compression and Union by Rank
class DisjointSet {
public:
    std::vector<int> parent;
    std::vector<int> rank;

    DisjointSet(int n) {
        parent.resize(n);
        std::iota(parent.begin(), parent.end(), 0); // Initialize each element as its own parent
        rank.assign(n, 0);
    }

    // Find operation with path compression
    int find(int i) {
        if (parent[i] == i) {
            return i;
        }
        return parent[i] = find(parent[i]);
    }

    // Union operation by rank
    bool unite(int i, int j) {
        int root_i = find(i);
        int root_j = find(j);

        if (root_i != root_j) {
            if (rank[root_i] < rank[root_j]) {
                parent[root_i] = root_j;
            } else if (rank[root_i] > rank[root_j]) {
                parent[root_j] = root_i;
            } else {
                parent[root_j] = root_i;
                rank[root_i]++;
            }
            return true; // Union successful
        }
        return false; // Already in the same set (forms a cycle)
    }
};

// Kruskal's algorithm to find the Minimum Spanning Tree
std::vector<Edge> kruskalsAlgorithm(int num_nodes, std::vector<Edge>& edges) {
    std::vector<Edge> mst;

    // 1. Sort all edges in non-decreasing order of their weight
    std::sort(edges.begin(), edges.end());

    DisjointSet dsu(num_nodes);

    // 2. Iterate through the sorted edges
    for (const auto& edge : edges) {
        // 3. If including this edge does not form a cycle, add it to MST
        if (dsu.unite(edge.u, edge.v)) {
            mst.push_back(edge);
        }

        // Optimization: If MST has num_nodes - 1 edges, we are done
        if (mst.size() == num_nodes - 1) {
            break;
        }
    }

    return mst;
}

int main() {
    // --- Example Usage ---

    // Example Graph 1
    int num_nodes1 = 4;
    std::vector<Edge> edges1 = {
        {0, 1, 10},
        {0, 2, 6},
        {0, 3, 5},
        {1, 3, 15},
        {2, 3, 4}
    };

    std::cout << "Graph 1 (Nodes: " << num_nodes1 << ", Edges: ";
    for(const auto& e : edges1) {
        std::cout << "(" << e.u << "-" << e.v << ", w=" << e.weight << ") ";
    }
    std::cout << ")\n";

    std::vector<Edge> mst1 = kruskalsAlgorithm(num_nodes1, edges1);

    std::cout << "MST Edges for Graph 1:\n";
    long long total_weight1 = 0;
    for (const auto& edge : mst1) {
        std::cout << "  " << edge.u << " -- " << edge.v << " (Weight: " << edge.weight << ")\n";
        total_weight1 += edge.weight;
    }
    std::cout << "Total MST Weight for Graph 1: " << total_weight1 << "\n";

    std::cout << "\n" << std::string(30, '-') << "\n\n";

    // Example Graph 2
    int num_nodes2 = 7; // Nodes 0 to 6
    std::vector<Edge> edges2 = {
        {0, 1, 28},
        {0, 5, 10},
        {1, 2, 16},
        {1, 6, 14},
        {2, 3, 12},
        {3, 4, 22},
        {3, 6, 18},
        {4, 5, 25},
        {4, 6, 24}
    };

    std::cout << "Graph 2 (Nodes: " << num_nodes2 << ", Edges: ";
    for(const auto& e : edges2) {
        std::cout << "(" << e.u << "-" << e.v << ", w=" << e.weight << ") ";
    }
    std::cout << ")\n";

    std::vector<Edge> mst2 = kruskalsAlgorithm(num_nodes2, edges2);

    std::cout << "MST Edges for Graph 2:\n";
    long long total_weight2 = 0;
    for (const auto& edge : mst2) {
        std::cout << "  " << edge.u << " -- " << edge.v << " (Weight: " << edge.weight << ")\n";
        total_weight2 += edge.weight;
    }
    std::cout << "Total MST Weight for Graph 2: " << total_weight2 << "\n";

    return 0;
}
```

```python
class Edge:
    def __init__(self, u, v, weight):
        self.u = u
        self.v = v
        self.weight = weight

    # Used for sorting edges by weight
    def __lt__(self, other):
        return self.weight < other.weight

    def __repr__(self):
        return f"({self.u}-{self.v}, w={self.weight})"


class DisjointSet:
    def __init__(self, num_nodes):
        self.parent = list(range(num_nodes))
        self.rank = [0] * num_nodes

    # Find the representative (root) of the set containing element i
    def find(self, i):
        if self.parent[i] == i:
            return i
        self.parent[i] = self.find(self.parent[i]) # Path compression
        return self.parent[i]

    # Unites the set containing element i and the set containing element j
    def union(self, i, j):
        root_i = self.find(i)
        root_j = self.find(j)

        if root_i != root_j:
            # Union by rank to keep the tree flat
            if self.rank[root_i] < self.rank[root_j]:
                self.parent[root_i] = root_j
            elif self.rank[root_i] > self.rank[root_j]:
                self.parent[root_j] = root_i
            else:
                self.parent[root_j] = root_i
                self.rank[root_i] += 1
            return True  # Union successful
        return False # Elements are already in the same set

def kruskals_algorithm(num_nodes, edges):
    # 1. Sort all edges in non-decreasing order of their weight
    edges.sort()

    mst = []
    dsu = DisjointSet(num_nodes)

    # 2. Iterate through the sorted edges
    for edge in edges:
        u, v, weight = edge.u, edge.v, edge.weight

        # 3. If including this edge does not form a cycle, add it to MST
        if dsu.union(u, v):
            mst.append(edge)

        # Optimization: If MST has num_nodes - 1 edges, we are done
        if len(mst) == num_nodes - 1:
            break

    return mst

# --- Example Usage ---

# Represent a graph with 4 nodes (0, 1, 2, 3) and 5 edges
# Example Graph (from common Kruskal's algorithm examples)
# Edge: (u, v, weight)

num_nodes1 = 4
edges1 = [
    Edge(0, 1, 10),
    Edge(0, 2, 6),
    Edge(0, 3, 5),
    Edge(1, 3, 15),
    Edge(2, 3, 4)
]

print(f"Graph 1 (Nodes: {num_nodes1}, Edges: {edges1})")
mst1 = kruskals_algorithm(num_nodes1, edges1)
print("MST Edges for Graph 1:")
total_weight1 = 0
for edge in mst1:
    print(f"  {edge.u} -- {edge.v} (Weight: {edge.weight})")
    total_weight1 += edge.weight
print(f"Total MST Weight for Graph 1: {total_weight1}")

print("\n" + "-"*30 + "\n")

# Another example graph
num_nodes2 = 7 # Nodes 0 to 6
edges2 = [
    Edge(0, 1, 28),
    Edge(0, 5, 10),
    Edge(1, 2, 16),
    Edge(1, 6, 14),
    Edge(2, 3, 12),
    Edge(3, 4, 22),
    Edge(3, 6, 18),
    Edge(4, 5, 25),
    Edge(4, 6, 24)
]

print(f"Graph 2 (Nodes: {num_nodes2}, Edges: {edges2})")
mst2 = kruskals_algorithm(num_nodes2, edges2)
print("MST Edges for Graph 2:")
total_weight2 = 0
for edge in mst2:
    print(f"  {edge.u} -- {edge.v} (Weight: {edge.weight})")
    total_weight2 += edge.weight
print(f"Total MST Weight for Graph 2: {total_weight2}")
```

***4. Write a program to implement the Bellman-Ford algorithm to find the shortest paths from a given source node to all other nodes in a graph c++ then python***

```python
%%writefile bellman_ford.cpp
#include <iostream>
#include <vector>
#include <limits> // For std::numeric_limits

// Define a large value to represent infinity
const int INF = std::numeric_limits<int>::max();

// Structure to represent an edge
struct Edge {
    int u, v, weight;
};

// Bellman-Ford algorithm implementation
void bellmanFord(int num_nodes, const std::vector<Edge>& edges, int source) {
    std::vector<int> dist(num_nodes, INF);
    dist[source] = 0;

    // Relax all edges (num_nodes - 1) times
    for (int i = 0; i < num_nodes - 1; ++i) {
        for (const auto& edge : edges) {
            int u = edge.u;
            int v = edge.v;
            int weight = edge.weight;

            // If the distance to u is not infinity and a shorter path to v is found
            if (dist[u] != INF && dist[u] + weight < dist[v]) {
                dist[v] = dist[u] + weight;
            }
        }
    }

    // Check for negative cycles
    bool negative_cycle_detected = false;
    for (const auto& edge : edges) {
        int u = edge.u;
        int v = edge.v;
        int weight = edge.weight;

        if (dist[u] != INF && dist[u] + weight < dist[v]) {
            negative_cycle_detected = true;
            break;
        }
    }

    std::cout << "Shortest paths from source " << source << ":\n";
    if (negative_cycle_detected) {
        std::cout << "  Graph contains a negative cycle! Shortest paths are not well-defined.\n";
    } else {
        for (int i = 0; i < num_nodes; ++i) {
            if (dist[i] == INF) {
                std::cout << "  Node " << i << ": unreachable\n";
            } else {
                std::cout << "  Node " << i << ": " << dist[i] << "\n";
            }
        }
    }
}

int main() {
    // Seed for random operations (not strictly needed for Bellman-Ford, but good practice for C++ main)
    // srand(time(0));

    // --- Example Graph 1 (No negative cycle) ---
    int num_nodes1 = 5; // Nodes 0 to 4
    std::vector<Edge> edges1 = {
        {0, 1, -1},
        {0, 2, 4},
        {1, 2, 3},
        {1, 3, 2},
        {1, 4, 2},
        {3, 2, 5},
        {3, 1, 1},
        {4, 3, -3}
    };
    int source1 = 0;

    std::cout << "\n--- Graph 1 (No negative cycle) ---\n";
    std::cout << "Nodes: " << num_nodes1 << ", Edges:\n";
    for (const auto& edge : edges1) {
        std::cout << "  " << edge.u << " -> " << edge.v << " (Weight: " << edge.weight << ")\n";
    }
    bellmanFord(num_nodes1, edges1, source1);

    // --- Example Graph 2 (With a negative cycle) ---
    int num_nodes2 = 4; // Nodes 0 to 3
    std::vector<Edge> edges2 = {
        {0, 1, 1},
        {1, 2, -1},
        {2, 3, -1},
        {3, 0, -1} // This creates a negative cycle 0-1-2-3-0 with total weight -2
    };
    int source2 = 0;

    std::cout << "\n--- Graph 2 (With a negative cycle) ---\n";
    std::cout << "Nodes: " << num_nodes2 << ", Edges:\n";
    for (const auto& edge : edges2) {
        std::cout << "  " << edge.u << " -> " << edge.v << " (Weight: " << edge.weight << ")\n";
    }
    bellmanFord(num_nodes2, edges2, source2);

    return 0;
}
```

```python
import math

def bellman_ford(num_nodes, edges, source):
    # Initialize distances from source to all other nodes as infinity
    # and distance to source as 0.
    dist = [math.inf] * num_nodes
    dist[source] = 0

    # Relax all edges num_nodes - 1 times.
    # A simple path from source to any other vertex can have at most num_nodes - 1 edges.
    for _ in range(num_nodes - 1):
        for u, v, weight in edges:
            if dist[u] != math.inf and dist[u] + weight < dist[v]:
                dist[v] = dist[u] + weight

    # Check for negative cycles.
    # If we can still relax an edge, then a negative cycle exists.
    negative_cycle_detected = False
    for u, v, weight in edges:
        if dist[u] != math.inf and dist[u] + weight < dist[v]:
            negative_cycle_detected = True
            break

    print(f"Shortest paths from source {source}:")
    if negative_cycle_detected:
        print("  Graph contains a negative cycle! Shortest paths are not well-defined.")
    else:
        for i in range(num_nodes):
            if dist[i] == math.inf:
                print(f"  Node {i}: unreachable")
            else:
                print(f"  Node {i}: {dist[i]}")

# --- Example Usage ---

# Example Graph 1 (No negative cycle)
num_nodes1 = 5 # Nodes 0 to 4
edges1 = [
    (0, 1, -1),
    (0, 2, 4),
    (1, 2, 3),
    (1, 3, 2),
    (1, 4, 2),
    (3, 2, 5),
    (3, 1, 1),
    (4, 3, -3)
]
source1 = 0

print("\n--- Graph 1 (No negative cycle) ---")
print(f"Nodes: {num_nodes1}, Edges:")
for u, v, weight in edges1:
    print(f"  {u} -> {v} (Weight: {weight})")
bellman_ford(num_nodes1, edges1, source1)

# Example Graph 2 (With a negative cycle)
num_nodes2 = 4 # Nodes 0 to 3
edges2 = [
    (0, 1, 1),
    (1, 2, -1),
    (2, 3, -1),
    (3, 0, -1) # This creates a negative cycle 0-1-2-3-0 with total weight -2
]
source2 = 0

print("\n--- Graph 2 (With a negative cycle) ---")
print(f"Nodes: {num_nodes2}, Edges:")
for u, v, weight in edges2:
    print(f"  {u} -> {v} (Weight: {weight})")
bellman_ford(num_nodes2, edges2, source2)
```

```python
%%writefile b_tree.cpp
#include <iostream>
#include <vector>
#include <algorithm> // For std::find, std::sort

// A B-Tree node
class BTreeNode {
public:
    std::vector<int> keys;      // An array of keys
    std::vector<BTreeNode*> children; // An array of child pointers
    int t;                     // Minimum degree (defines the range for number of keys)
    bool leaf;                 // Is true when node is leaf. Otherwise false

    // Constructor
    BTreeNode(int t_val, bool leaf_val);

    // Function to traverse all nodes in a subtree rooted with this node
    void traverse();

    // Function to search key in this node
    BTreeNode* search(int k); // Returns NULL if k is not present.

    // A function that inserts a new key in this node
    void insertNonFull(int k);

    // A function that splits the child y of this node.
    // Note that y must be full when this function is called.
    void splitChild(int i, BTreeNode* y);

    friend class BTree;
};

// A B-Tree
class BTree {
public:
    BTreeNode* root; // Pointer to root node
    int t;           // Minimum degree

    // Constructor (Initializes an empty B-tree)
    BTree(int t_val) {
        root = nullptr;
        t = t_val;
    }

    void traverse() {
        if (root != nullptr) root->traverse();
        std::cout << std::endl;
    }

    // Function to search key in this tree
    BTreeNode* search(int k) {
        return (root == nullptr) ? nullptr : root->search(k);
    }

    // The main function that inserts a new key in this B-Tree
    void insert(int k);
};

// Constructor for BTreeNode
BTreeNode::BTreeNode(int t_val, bool leaf_val) {
    t = t_val;
    leaf = leaf_val;
    keys.reserve(2 * t - 1);
    children.reserve(2 * t);
}

// Function to traverse all nodes in a subtree rooted with this node
void BTreeNode::traverse() {
    int i;
    for (i = 0; i < keys.size(); i++) {
        // If this is not leaf, then before printing key[i], traverse the subtree rooted with child C[i].
        if (leaf == false) {
            children[i]->traverse();
        }
        std::cout << " " << keys[i];
    }

    // Print the subtree rooted with last child
    if (leaf == false) {
        children[i]->traverse();
    }
}

// Function to search key in this node
BTreeNode* BTreeNode::search(int k) {
    // Find the first key greater than or equal to k
    int i = 0;
    while (i < keys.size() && k > keys[i]) {
        i++;
    }

    // If the found key is equal to k, return this node
    if (i < keys.size() && keys[i] == k) {
        return this;
    }

    // If key is not found here and this is a leaf node
    if (leaf == true) {
        return nullptr;
    }

    // Go to the appropriate child
    return children[i]->search(k);
}

// The main function that inserts a new key in this B-Tree
void BTree::insert(int k) {
    // If tree is empty
    if (root == nullptr) {
        // Allocate memory for root
        root = new BTreeNode(t, true);
        root->keys.push_back(k); // Insert key
    } else { // If tree is not empty
        // If root is full, then tree grows in height
        if (root->keys.size() == 2 * t - 1) {
            // Allocate memory for new root
            BTreeNode* s = new BTreeNode(t, false);

            // Make old root as first child of new root
            s->children.push_back(root);

            // Split the old root and move 1 key to the new root
            s->splitChild(0, root);

            // New root has two children now. Decide which of the two children is going to have new key
            int i = 0;
            if (s->keys[0] < k)
                i++;
            s->children[i]->insertNonFull(k);

            // Change root
            root = s;
        } else { // If root is not full, call insertNonFull for root
            root->insertNonFull(k);
        }
    }
}

// A function that inserts a new key in this node
// The node must be non-full when this function is called
void BTreeNode::insertNonFull(int k) {
    // Initialize index as rightmost element
    int i = keys.size() - 1;

    // If this is a leaf node
    if (leaf == true) {
        // Find where to insert the new key
        // All elements to the right of 'i' are shifted one place ahead
        keys.push_back(0); // Make space for new key
        while (i >= 0 && keys[i] > k) {
            keys[i + 1] = keys[i];
            i--;
        }
        keys[i + 1] = k; // Insert the new key
    } else { // If this node is not leaf
        // Find the child which is going to have the new key
        while (i >= 0 && keys[i] > k) {
            i--;
        }
        i++;

        // See if the identified child is full
        if (children[i]->keys.size() == 2 * t - 1) {
            // If the child is full, then split it
            splitChild(i, children[i]);

            // After splitting, the middle key of children[i] goes up and
            // children[i] is split into two. See which of the two is going to have the new key
            if (keys[i] < k)
                i++;
        }
        children[i]->insertNonFull(k);
    }
}

// A function that splits the child y of this node.
// i is index of y in child array C[] of this node.
// The child y must be full when this function is called
void BTreeNode::splitChild(int i, BTreeNode* y) {
    // Create a new node which is going to store (t-1) keys
    // and t children of y
    BTreeNode* z = new BTreeNode(y->t, y->leaf);

    // Copy the last (t-1) keys of y to z
    for (int j = 0; j < t - 1; j++) {
        z->keys.push_back(y->keys[j + t]);
    }

    // Copy the last t children of y to z
    if (y->leaf == false) {
        for (int j = 0; j < t; j++) {
            z->children.push_back(y->children[j + t]);
        }
    }

    // Reduce the number of keys in y
    y->keys.erase(y->keys.begin() + t - 1, y->keys.end());
    if (y->leaf == false) {
        y->children.erase(y->children.begin() + t, y->children.end());
    }

    // Since this node is going to have a new child, create space of new child
    children.insert(children.begin() + i + 1, z);

    // A key of y is going to move to this node. Find location of new key and insert it
    keys.insert(keys.begin() + i, y->keys[t - 1]);
}

int main() {
    // A B-Tree with minimum degree 3
    BTree t(3);

    std::cout << "Inserting keys 10, 20, 5, 6, 12, 30, 7, 17\n";
    t.insert(10);
    t.insert(20);
    t.insert(5);
    t.insert(6);
    t.insert(12);
    t.insert(30);
    t.insert(7);
    t.insert(17);

    std::cout << "Traversal of the constructed B-Tree is: ";
    t.traverse();

    int k = 6;
    (t.search(k) != nullptr) ? std::cout << "Present\n" : std::cout << "Not Present\n";

    k = 15;
    (t.search(k) != nullptr) ? std::cout << "Present\n" : std::cout << "Not Present\n";

    std::cout << "\nInserting more keys: 2, 4, 18, 25, 3, 16, 22, 24, 26, 19, 1, 21, 23, 11, 27\n";
    t.insert(2);
    t.insert(4);
    t.insert(18);
    t.insert(25);
    t.insert(3);
    t.insert(16);
    t.insert(22);
    t.insert(24);
    t.insert(26);
    t.insert(19);
    t.insert(1);
    t.insert(21);
    t.insert(23);
    t.insert(11);
    t.insert(27);

    std::cout << "Traversal of the B-Tree after more insertions is: ";
    t.traverse();

    k = 23;
    (t.search(k) != nullptr) ? std::cout << "Present\n" : std::cout << "Not Present\n";
    k = 99;
    (t.search(k) != nullptr) ? std::cout << "Present\n" : std::cout << "Not Present\n";

    return 0;
}
```

**6. Write a program to search for a pattern in a given text using the KMP algorithm in c++**

```python
%%writefile kmp_search.cpp
#include <iostream>
#include <vector>
#include <string>

// Function to compute the Longest Proper Prefix which is also a Suffix (LPS) array
// for the given pattern.
// The LPS array stores the length of the longest proper prefix of pattern[0...q]
// that is also a suffix of pattern[0...q].
void computeLPSArray(const std::string& pattern, std::vector<int>& lps) {
    int length = 0; // Length of the previous longest prefix suffix
    lps[0] = 0; // lps[0] is always 0
    int i = 1;
    int m = pattern.length();

    // The loop calculates lps[i] for i = 1 to m-1
    while (i < m) {
        if (pattern[i] == pattern[length]) {
            length++;
            lps[i] = length;
            i++;
        } else { // (pattern[i] != pattern[length])
            // This is tricky. Consider the example.
            // AAACAAAA and i = 7. The char pattern[i] is 'A'.
            // length is 3. The char pattern[length] is 'C'.
            // Since there is a mismatch, and length is not 0,
            // we set length = lps[length-1].
            // The idea is to not match prefix with prefix of pattern,
            // but rather to try matching smaller suffix with prefix of pattern.
            if (length != 0) {
                length = lps[length - 1];
            } else { // if (length == 0)
                lps[i] = 0;
                i++;
            }
        }
    }
}

// KMP algorithm to search for occurrences of a pattern in a text
void KMPSearch(const std::string& text, const std::string& pattern) {
    int n = text.length();
    int m = pattern.length();

    if (m == 0) {
        std::cout << "Pattern is empty. No search performed.\n";
        return;
    }
    if (n == 0 || m > n) {
        std::cout << "Pattern not found (text too short or empty).\n";
        return;
    }

    // Create LPS array to store longest proper prefix suffix values
    std::vector<int> lps(m);
    computeLPSArray(pattern, lps);

    int i = 0; // index for text
    int j = 0; // index for pattern

    while (i < n) {
        if (pattern[j] == text[i]) {
            i++;
            j++;
        }

        if (j == m) {
            std::cout << "Found pattern at index " << i - j << std::endl;
            j = lps[j - 1]; // To find next occurrences
        } else if (i < n && pattern[j] != text[i]) {
            // Do not match lps[0...lps[j-1]] characters,
            // they will match anyway
            if (j != 0) {
                j = lps[j - 1];
            } else {
                i++;
            }
        }
    }
}

int main() {
    // Example 1
    std::string text1 = "ABABDABACDABABCABAB";
    std::string pattern1 = "ABABCABAB";
    std::cout << "Text: " << text1 << std::endl;
    std::cout << "Pattern: " << pattern1 << std::endl;
    KMPSearch(text1, pattern1);
    std::cout << std::endl;

    // Example 2
    std::string text2 = "AAAAAA";
    std::string pattern2 = "AAA";
    std::cout << "Text: " << text2 << std::endl;
    std::cout << "Pattern: " << pattern2 << std::endl;
    KMPSearch(text2, pattern2);
    std::cout << std::endl;

    // Example 3 (No match)
    std::string text3 = "HELLO WORLD";
    std::string pattern3 = "GOODBYE";
    std::cout << "Text: " << text3 << std::endl;
    std::cout << "Pattern: " << pattern3 << std::endl;
    KMPSearch(text3, pattern3);
    std::cout << std::endl;

    // Example 4 (Pattern at beginning)
    std::string text4 = "TESTING KMP ALGORITHM";
    std::string pattern4 = "TESTING";
    std::cout << "Text: " << text4 << std::endl;
    std::cout << "Pattern: " << pattern4 << std::endl;
    KMPSearch(text4, pattern4);
    std::cout << std::endl;

    // Example 5 (Pattern at end)
    std::string text5 = "KMP ALGORITHM TESTING";
    std::string pattern5 = "TESTING";
    std::cout << "Text: " << text5 << std::endl;
    std::cout << "Pattern: " << pattern5 << std::endl;
    KMPSearch(text5, pattern5);
    std::cout << std::endl;

    return 0;
}
```

```python
def compute_lps_array(pattern):
    m = len(pattern)
    lps = [0] * m
    length = 0 # Length of the previous longest prefix suffix
    i = 1

    # The loop calculates lps[i] for i = 1 to m-1
    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1
    return lps

def kmp_search(text, pattern):
    n = len(text)
    m = len(pattern)

    if m == 0:
        print("Pattern is empty. No search performed.")
        return
    if n == 0 or m > n:
        print("Pattern not found (text too short or empty).")
        return

    lps = compute_lps_array(pattern)

    i = 0 # index for text
    j = 0 # index for pattern
    found_indices = []

    while i < n:
        if pattern[j] == text[i]:
            i += 1
            j += 1

        if j == m:
            found_indices.append(i - j)
            j = lps[j - 1] # To find next occurrences
        elif i < n and pattern[j] != text[i]:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1

    if not found_indices:
        print("Pattern not found.")
    else:
        for index in found_indices:
            print(f"Found pattern at index {index}")

# --- Example Usage ---

print("Example 1:")
text1 = "ABABDABACDABABCABAB"
pattern1 = "ABABCABAB"
print(f"Text: {text1}")
print(f"Pattern: {pattern1}")
kmp_search(text1, pattern1)
print()

print("Example 2:")
text2 = "AAAAAA"
pattern2 = "AAA"
print(f"Text: {text2}")
print(f"Pattern: {pattern2}")
kmp_search(text2, pattern2)
print()

print("Example 3 (No match):")
text3 = "HELLO WORLD"
pattern3 = "GOODBYE"
print(f"Text: {text3}")
print(f"Pattern: {pattern3}")
kmp_search(text3, pattern3)
print()

print("Example 4 (Pattern at beginning):")
text4 = "TESTING KMP ALGORITHM"
pattern4 = "TESTING"
print(f"Text: {text4}")
print(f"Pattern: {pattern4}")
kmp_search(text4, pattern4)
print()

print("Example 5 (Pattern at end):")
text5 = "KMP ALGORITHM TESTING"
pattern5 = "TESTING"
print(f"Text: {text5}")
print(f"Pattern: {pattern5}")
kmp_search(text5, pattern5)
print()
```

```python
%%writefile suffix_tree.cpp
#include <iostream>
#include <string>
#include <vector>
#include <map>
#include <algorithm> // For std::sort and std::unique
#include <set>       // For unique indices without sorting

// A node in the Suffix Tree
class SuffixNode {
public:
    std::map<char, SuffixNode*> children; // Pointers to child nodes, keyed by character
    std::set<int> suffix_indices;        // Stores starting indices of suffixes passing through this node

    SuffixNode() {}

    // Destructor to free allocated child nodes
    ~SuffixNode() {
        for (auto const& [key, val] : children) {
            delete val;
        }
    }
};

// Brute Force Suffix Tree implementation
class BruteForceSuffixTree {
public:
    SuffixNode* root;
    std::string original_text; // Store original text for reference, without terminator

    BruteForceSuffixTree(const std::string& text) {
        root = new SuffixNode();
        original_text = text;
        _build_suffix_tree();
    }

    // Destructor to free the root node and its children
    ~BruteForceSuffixTree() {
        delete root;
    }

private:
    // Helper function to insert a suffix into the tree
    void _insert_suffix(const std::string& suffix, int start_index) {
        SuffixNode* current_node = root;
        current_node->suffix_indices.insert(start_index);

        for (char ch : suffix) {
            if (current_node->children.find(ch) == current_node->children.end()) {
                current_node->children[ch] = new SuffixNode();
            }
            current_node = current_node->children[ch];
            current_node->suffix_indices.insert(start_index);
        }
    }

    // Builds the suffix tree by inserting all suffixes of the text
    void _build_suffix_tree() {
        // Append a unique terminator character to ensure all suffixes are distinct
        // and prevent issues where one suffix is a prefix of another.
        std::string terminated_text = original_text + '$';

        for (int i = 0; i < terminated_text.length(); ++i) {
            _insert_suffix(terminated_text.substr(i), i);
        }
    }

public:
    // Searches for a pattern in the suffix tree
    std::vector<int> search(const std::string& pattern) {
        SuffixNode* current_node = root;
        for (char ch : pattern) {
            if (current_node->children.find(ch) == current_node->children.end()) {
                return {}; // Pattern not found
            }
            current_node = current_node->children[ch];
        }
        // Convert set to vector and return sorted unique indices
        std::vector<int> results(current_node->suffix_indices.begin(), current_node->suffix_indices.end());
        std::sort(results.begin(), results.end());
        // Remove any indices that are for suffixes of the terminator character itself and not the original text.
        // The terminator character '$' is always at `original_text.length()`. Any suffix index equal to or greater
        // than this means it's part of the terminator or a suffix starting exactly at the terminator.
        results.erase(std::remove_if(results.begin(), results.end(),
                                     [this](int idx) { return idx >= original_text.length(); }),
                      results.end());
        return results;
    }

    // Helper function to print the tree structure (for debugging/visualization)
    void print_tree(SuffixNode* node, int indent = 0, char edge_char = 'R') const {
        for (int i = 0; i < indent; ++i) {
            std::cout << "  ";
        }
        std::cout << "[" << edge_char << "] (Indices: ";
        bool first = true;
        for (int idx : node->suffix_indices) {
             if (idx < original_text.length()) { // Only print indices for the original text
                if (!first) std::cout << ", ";
                std::cout << idx;
                first = false;
            }
        }
        std::cout << ")\n";

        for (auto const& [ch, child_node] : node->children) {
            print_tree(child_node, indent + 1, ch);
        }
    }
};

int main() {
    // --- Example Usage 1 ---
    std::string text1 = "banana";
    std::cout << "Building suffix tree for text: '" << text1 << "'\n";
    BruteForceSuffixTree st1(text1);
    std::cout << "Suffix tree structure (showing relevant indices):\n";
    st1.print_tree(st1.root);

    std::cout << "\n--- Searching patterns for '" << text1 << "' ---\n";
    std::vector<std::string> patterns_to_search1 = {"ana", "ban", "na", "a", "bana", "nana", "x"};
    for (const std::string& pattern : patterns_to_search1) {
        std::vector<int> occurrences = st1.search(pattern);
        if (!occurrences.empty()) {
            std::cout << "Pattern '" << pattern << "' found at indices: ";
            for (int idx : occurrences) {
                std::cout << idx << " ";
            }
            std::cout << "\n";
        } else {
            std::cout << "Pattern '" << pattern << "' not found.\n";
        }
    }

    std::cout << "\n" << std::string(40, '=') << "\n\n";

    // --- Example Usage 2 ---
    std::string text2 = "abracadabra";
    std::cout << "Building suffix tree for text: '" << text2 << "'\n";
    BruteForceSuffixTree st2(text2);
    std::cout << "Suffix tree structure (showing relevant indices):\n";
    st2.print_tree(st2.root);

    std::cout << "\n--- Searching patterns for '" << text2 << "' ---\n";
    std::vector<std::string> patterns_to_search2 = {"abra", "cad", "a", "bracad", "z"};
    for (const std::string& pattern : patterns_to_search2) {
        std::vector<int> occurrences = st2.search(pattern);
        if (!occurrences.empty()) {
            std::cout << "Pattern '" << pattern << "' found at indices: ";
            for (int idx : occurrences) {
                std::cout << idx << " ";
            }
            std::cout << "\n";
        } else {
            std::cout << "Pattern '" << pattern << "' not found.\n";
        }
    }

    return 0;
}
```

```python
class SuffixNode:
    def __init__(self):
        self.children = {}
        self.suffix_indices = [] # Stores starting indices of suffixes passing through this node

class BruteForceSuffixTree:
    def __init__(self, text):
        self.root = SuffixNode()
        self.text = text
        self._build_suffix_tree()

    def _insert_suffix(self, suffix, start_index):
        current_node = self.root
        current_node.suffix_indices.append(start_index)
        for char in suffix:
            if char not in current_node.children:
                current_node.children[char] = SuffixNode()
            current_node = current_node.children[char]
            current_node.suffix_indices.append(start_index)

    def _build_suffix_tree(self):
        # Add a unique terminator character to the text
        # This helps ensure that no suffix is a prefix of another suffix
        # for proper suffix tree construction and pattern matching.
        terminated_text = self.text + '$'
        for i in range(len(terminated_text)):
            suffix = terminated_text[i:]
            self._insert_suffix(suffix, i)

    def search(self, pattern):
        current_node = self.root
        for char in pattern:
            if char in current_node.children:
                current_node = current_node.children[char]
            else:
                return [] # Pattern not found
        # If we reached here, the pattern exists in the tree
        # The suffix_indices at this node represent all occurrences of the pattern
        return sorted(list(set(current_node.suffix_indices))) # Return unique and sorted indices

    def print_tree(self, node, indent=0, char='ROOT'):
        print('  ' * indent + f"[{char}] (Indices: {sorted(list(set(node.suffix_indices)))})")
        for c, child in node.children.items():
            self.print_tree(child, indent + 1, c)

# --- Example Usage ---

text1 = "banana"
print(f"Building suffix tree for text: '{text1}'")
suffix_tree1 = BruteForceSuffixTree(text1)
print("Suffix tree structure:")
suffix_tree1.print_tree(suffix_tree1.root)

print("\n--- Searching patterns ---")
patterns_to_search1 = ["ana", "ban", "na", "a", "bana", "nana", "x"]
for pattern in patterns_to_search1:
    occurrences = suffix_tree1.search(pattern)
    if occurrences:
        print(f"Pattern '{pattern}' found at indices: {occurrences}")
    else:
        print(f"Pattern '{pattern}' not found.")

print("\n" + "="*40 + "\n")

text2 = "abracadabra"
print(f"Building suffix tree for text: '{text2}'")
suffix_tree2 = BruteForceSuffixTree(text2)
print("Suffix tree structure:")
suffix_tree2.print_tree(suffix_tree2.root)

print("\n--- Searching patterns ---")
patterns_to_search2 = ["abra", "cad", "a", "bracad", "z"]
for pattern in patterns_to_search2:
    occurrences = suffix_tree2.search(pattern)
    if occurrences:
        print(f"Pattern '{pattern}' found at indices: {occurrences}")
    else:
        print(f"Pattern '{pattern}' not found.")
```
